```python
import trimesh
import open3d as o3d
import numpy as np

def trimesh_to_open3d(mesh):
    o3d_mesh = o3d.geometry.TriangleMesh()
    o3d_mesh.vertices = o3d.utility.Vector3dVector(mesh.vertices)
    o3d_mesh.triangles = o3d.utility.Vector3iVector(mesh.faces)
    return o3d_mesh


def open3d_to_trimesh(o3d_mesh):
    vertices = np.asarray(o3d_mesh.vertices)
    faces = np.asarray(o3d_mesh.triangles)
    return trimesh.Trimesh(vertices=vertices, faces=faces)


def normalize_mesh(mesh):
    mesh = mesh.copy()
    mesh.apply_translation(-mesh.centroid)
    scale = max(mesh.extents)
    if scale > 0:
        mesh.apply_scale(1.0 / scale)
    return mesh


def load_mesh(path):
    mesh = trimesh.load(path)
    if isinstance(mesh, trimesh.Scene):
        mesh = trimesh.util.concatenate(mesh.dump())
    return mesh


mesh1 = load_mesh("../input-data/external-atlas/brain-female-julich/julich-brain.glb")
mesh2 = load_mesh("../raw-data/external-atlas/brain-female-julich/3d-allen-f-brain.glb")

mesh1 = normalize_mesh(mesh1)
mesh2 = normalize_mesh(mesh2)

o3d_mesh1 = trimesh_to_open3d(mesh1)

v1 = len(mesh1.vertices)
v2 = len(mesh2.vertices)

print("Mesh1 vertices:", v1)
print("Mesh2 vertices:", v2)

target_vertices = v2  

if v1 > v2:
    print("Downsampling Mesh1...")
    target_triangles = int(len(mesh1.faces) * (v2 / v1))
    o3d_mesh1 = o3d_mesh1.simplify_quadric_decimation(target_triangles)

elif v1 < v2:
    print("Upsampling Mesh1...")
    while len(o3d_mesh1.vertices) < v2:
        o3d_mesh1 = o3d_mesh1.subdivide_midpoint(1)

else:
    print("Vertex count already matches")

def clean_mesh(o3d_mesh):
    o3d_mesh.remove_degenerate_triangles()
    o3d_mesh.remove_duplicated_triangles()
    o3d_mesh.remove_duplicated_vertices()
    o3d_mesh.remove_non_manifold_edges()
    return o3d_mesh

o3d_mesh1 = clean_mesh(o3d_mesh1)

final_vertices = np.asarray(o3d_mesh1.vertices)

if len(final_vertices) > v2:
    indices = np.random.choice(len(final_vertices), v2, replace=False)
    final_vertices = final_vertices[indices]

o3d_mesh1.vertices = o3d.utility.Vector3dVector(final_vertices)

mesh1_matched = open3d_to_trimesh(o3d_mesh1)

mesh1_matched.export("../input-data/external-atlas/brain-female-julich/julich-brain.glb")

print("Mesh1 updated to match Mesh2 vertex count!")
```
